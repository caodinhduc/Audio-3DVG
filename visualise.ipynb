```python
# scene_id = 'scene0125_00'
scene_id = 'scene0050_00'
PATH = '/home/duccd/Desktop/InstanceRefer/data/scannet/pointgroup_data'
mesh_vertices = np.load(os.path.join(PATH, scene_id) + "_aligned_vert.npy")  # axis-aligned
instance_labels = np.load(os.path.join(PATH, scene_id) + "_ins_label.npy")
semantic_labels = np.load(os.path.join(PATH, scene_id) + "_sem_label_pg.npy")
instance_bboxes = np.load(os.path.join(PATH, scene_id) + "_aligned_bbox.npy")
```

```python
import open3d as o3d

# Create a sample point cloud (random points)
points = mesh_vertices[:, 0:3]
pcl_color = mesh_vertices[:, 3:6]/255.0
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)
pcd.colors = o3d.utility.Vector3dVector(pcl_color)

# Define multiple axis-aligned bounding boxes
aabbs = []
id=0
for box in instance_bboxes:
    center = box[0:3].tolist()
    size = box[3:6].tolist()
    min_bound = np.array(center) - np.array(size) / 2
    max_bound = np.array(center) + np.array(size) / 2
    aabb = o3d.geometry.AxisAlignedBoundingBox(min_bound, max_bound)
    aabb.color = (0, 1, 0)  # green
    if id==0:
        aabbs.append(aabb)
    id+= 1
# Visualize point cloud with bounding boxes
o3d.visualization.draw_geometries([pcd])
```
